# The aim of this notebook is to test the SLIC superpixel segmentation algorithm on sample images from the dataset. 

```python
from skimage.segmentation import slic, mark_boundaries
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
```

```python
def slic_segment(image, n_segments=100, compactness=10):
    image_np = np.array(image)
    
    # Apply SLIC algorithm
    segments = slic(image_np, n_segments=n_segments, compactness=compactness, start_label=1)
    
    # Mark boundaries on the original image
    segmented_image = mark_boundaries(image_np, segments)

    segmented_image = (segmented_image * 255).astype(np.uint8)
    segmented_image = Image.fromarray(segmented_image)
    
    return segmented_image, segments
```

# Test the SLIC segementation algorithm on a random image with different number of segments

```python
ssd_path = "/Volumes/PortableSSD/PDS_eugenie"

image_path = f"{ssd_path}/Fonds 2021/10-13 ans 2021/2021_10-13_0539_TWN_R_C.jpg"
image = Image.open(image_path)


print("Number of segments: 50")
segmented_image, segments = slic_segment(image, n_segments=50)
display(segmented_image)


print("Number of segments: 500")
segmented_image, segments = slic_segment(image, n_segments=500)
display(segmented_image)
```
