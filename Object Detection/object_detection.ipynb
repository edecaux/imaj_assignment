# The aim of this notebook is to test if an object detection model can detect objects in drawings.

```python
import torch

# Load YOLOv5 model 
model = torch.hub.load("ultralytics/yolov5", "yolov5s")  
```

# Testing Pre-trained YOLOv5 on the Dataset

```python
ssd_path = "/Volumes/PortableSSD/PDS_eugenie"
image_path2 = f"{ssd_path}/Fonds 2021/10-13 ans 2021/2021_10-13_0548_CHI_R_C.jpg"
img = image_path2

results = model(img)

results.print()  
results.show()  
results.save()  


image_path3 = f"{ssd_path}/Fonds 2021/3-5 ans 2021/2021_3-5_0906_THA_R_C.jpg"
img = image_path3

results = model(img)

results.print()  
results.show()  
results.save()  
```

# Testing fine-tuned Yolov5 on the water theme year

```python
from ultralytics import YOLO
model = YOLO('runs/detect/train/weights/best.pt')
model100 = YOLO('runs/detect/train/weights100/weights/best.pt')
model1000 = YOLO('runs/detect/train/weights1000/weights/best.pt')
```

Run the cells below to see the results

```python
ssd_path = "/Volumes/PortableSSD/PDS_eugenie"

# No fish
image_path = f"{ssd_path}/Fonds 2017/6-9 ans 2017/2017_6-9_0743_RUS_R_C.jpg"

results = model(image_path)
results[0].show()

results100 = model100(image_path)
results100[0].show()

results1000 = model1000(image_path)
results1000[0].show()
```

```python
# Image with fish
image_path = f"{ssd_path}/Fonds 2017/6-9 ans 2017/2017_6-9_1868_ROM_R_C.jpg"
results = model(image_path)
results[0].show()

results100 = model100(image_path)
results100[0].show()

results1000 = model1000(image_path)
results1000[0].show()
```
